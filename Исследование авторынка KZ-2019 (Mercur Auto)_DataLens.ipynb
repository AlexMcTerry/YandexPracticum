### Обработка столбца fuel

<span style="font-size:16px">Примем следущую систему обозначения:<br> D - disel(Дизель)<br>
P - petrol(Бензин)<br>
H - hybrid(Гибрид)<br>
E - electro(Электрокар)</span>

```python
# приводим все виды топлива к нижнему регистру
data['fuel'] = data['fuel'].str.lower()
# т.к. там все +/- одинаково, меняем только электро/электрическая на E
data.loc[data['fuel'].str.contains('элект',
                                              na=False,
                                              regex=True),'fuel'] = 'E'
```

```python
# перекодируем виды топлива
data['fuel'] = data['fuel'].str.replace('бензин','P')
data['fuel'] = data['fuel'].str.replace('дизель','D')
data['fuel'] = data['fuel'].str.replace('гибрид','H')
```

```python
# в таблице есть сдвинутые колонки. Исправляем эту ситуацию
data.loc[data['fuel'].str.contains('2|1,6', na=False, regex=True),'transmission_type'] = data.loc[data['fuel'].str.contains('2|1,6', na=False, regex=True),'engine_capacity_liters']
data.loc[data['fuel'].str.contains('2|1,6', na=False, regex=True),'engine_capacity_liters'] = data.loc[data['fuel'].str.contains('2|1,6', na=False, regex=True),'fuel']
# получившийся пробел в виде топлива заполняется NaN
data.loc[data['fuel'].str.contains('2|1,6', na=False, regex=True),'fuel'] = np.nan
data.loc[data['fuel'].str.contains('0', na=False, regex=True),'fuel'] = np.nan
```

```python
# находим модели, у которых присутствуют NaN 
wout_fuel = data.loc[~data['fuel'].str.contains('P|D|H|E',na=False,regex=True)]['model'].unique()
# для каждой модели считаем количесвто вариантов топлива
for i in wout_fuel:
    if (data.query('model == @i')['fuel'].nunique() == 1):
        # если во всех остальных строках для этой модели указана только один тип топлива, то его и прописываем для значений NaN
        fuel = data.loc[(data['model']==i) & (data['fuel'].notna()),'fuel'].values[0]
        data.loc[data['model'] == i,'fuel'] = fuel
    if (data.query('model == @i')['fuel'].nunique() > 1):
        # если больше одного типа топлива, то для пробелов указываем 'Не указано'
        data.loc[(data['model'] == i)&(data['fuel'].isna()),'fuel'] = 'Не указано'
```
